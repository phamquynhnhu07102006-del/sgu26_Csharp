## Bài 2: Xuất và nhập chuỗi
Dùng chương trình Visual Studio.Net viết chương trình nhập họ tên và xuất họ tên đã nhập ra màn hình
console theo định dạng sau:

```
using Microsoft.DotNet.Interactive;

Console.Write("Nhap ho ten cua ban: ");

var hoTen = await Kernel.GetInputAsync("Ho ten:");

Console.WriteLine("Chao ban " + hoTen + "!");
```
